# Handling Outliers

```python

titanic_df.Age.plot.box()
```

```python
Q1 = titanic_df['Age'].quantile(0.25)
print(f"Q1 :{Q1}")
Q3 = titanic_df['Age'].quantile(0.75)
print(f"Q3 :{Q3}")
IQR = Q3 - Q1
print(f"IQR: {IQR}")
upper_limit = Q3 + 1.5 * IQR
lower_limit = Q1 - 1.5 * IQR
```

```python
titanic_df[(titanic_df['Age'] < upper_limit) & (titanic_df['Age'] > lower_limit) ]
```
