### **Problem 4: Numpy Array Operations**  
1. Convert the numerical columns (such as `LotArea`, `YearBuilt`, `GrLivArea`, `SalePrice`, etc.) into a **NumPy array**.
2. Perform the following NumPy operations on the array:
   - Compute the **mean** and **standard deviation** of each column.
   - Normalize the data using **min-max scaling** and **z-score normalization**.
   - Compute the **dot product** between two numerical columns of your choice.

```python
import pandas as pd
import numpy as np

numerical_columns = df.select_dtypes(include = ['int64','float64']).columns

numpy_numerical_array = df[numerical_columns].to_numpy()

numpy_numerical_array
```

```python
numpy_mean = np.mean(numpy_numerical_array, axis = 0)
print('Mean: ',numpy_mean)

numpy_std = np.std(numpy_numerical_array, axis = 0)
print("\n Standard Deviation: ",numpy_std)
```

```python
# Normalize the data using min-max scaling and z-score normalization.

# Min-Max Scaling
min_vals = np.min(numpy_numerical_array, axis=0)
max_vals = np.max(numpy_numerical_array, axis=0)

min_max_scaled_array = (numpy_numerical_array - min_vals) / (max_vals - min_vals)


# Display Min-Max Scaled Data
print("Min-Max Scaled Data:\n", min_max_scaled_array)
```

```python
# Z score normalization

z_score = (numpy_numerical_array - numpy_mean)/ numpy_std

print("Z-Score Normalized Data:\n", z_score)
```

### **Problem 6: Aggregation and Transformation**  
1. Perform a **groupby** operation on `Neighborhood` and calculate the **mean** and **median** for `SalePrice`.
2. Use the `apply()` function to create a custom transformation for `SalePrice` (e.g., scaling or transformation).
3. Aggregate multiple statistical values (e.g., **mean**, **median**, and **count**) for `SalePrice` grouped by `OverallQual`.

```python
groupby_mean = df.groupby('Neighborhood')['SalePrice'].mean()
groupby_median = df.groupby('Neighborhood')['SalePrice'].median()

print(f'Mean : {groupby_mean}')
print(f'\n \n Median : {groupby_median}')
```

```python
df['house_category'] = df['SalePrice'].apply(lambda x: 
    'Luxury' if x > 300000 
    else 'Mid_Range' if x > 150000 and x <= 300000
    else 'Affordable')
```

```python
df['house_category']
```

```python
# Aggregate multiple statistical values (e.g., mean, median, and count) for SalePrice grouped by OverallQual.

groupby_overall = df.groupby('OverallQual').agg({'SalePrice': ['mean','median','count']})
groupby_overall
```
