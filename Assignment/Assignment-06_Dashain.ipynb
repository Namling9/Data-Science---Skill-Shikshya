### **Problem 1: Numerical Operations and String Manipulation**  
Write a Python program that:

````python
#importing math module
import math

class Math :

    # Creating power function
    def power(self,a,b):
        return a**b

    #Creating modulus function
    def modulus(self,a,b):
        return a%b

    #Creating floor_and_ceiling function to find both floor and ceil at the same time
    def floor_and_ceiling(self,a,b):
        print(f"floor of a:{math.floor(a)}, floor of b : {math.floor(b)}")
        print(f"ceiling of a:{math.ceil(a)} ceiling of b : {math.ceil(b)}")

math_instance = Math()

print(math_instance.power(2,3))
print(math_instance.modulus(6,87))
math_instance.floor_and_ceiling(3.23,64.23)
````

````python
# Taking string input from the user.
input_str = input("Enter the string: ")

# Creating a dictionary of frequency and the letter of the given string
frequency_count = { i : input_str.count(i) for i in input_str}
print(frequency_count)

print('Highest frequecy : ', max(frequency_count))

def find_most_frequent_character(frequency_dict):
    # Find the character with the maximum frequency
    most_frequent_char = max(frequency_dict, key=frequency_dict.get)
    return most_frequent_char, frequency_dict[most_frequent_char]

most_frequent_char, frequency = find_most_frequent_character(frequency_count)
print(f"\nThe character that occurs the most is '{most_frequent_char}' with a frequency of {frequency}.")

words = input_str.split()
acronym = ''.join(word[0].upper() for word in words)
print("Acronym: ",acronym)
````

Write a program that:

2. Calculates the **total sales** for each month and stores the result in a new dictionary.
3. Identifies the **best performing month** by total sales.
4. Extracts and prints the top **two categories** that contributed the most to the sales in the best performing month.

````python
import random
sales = {
    'January': {'Electronics': 5000, 'Clothing': 3000, 'Grocery': 2000},
    'February': {'Electronics': 6000, 'Clothing': 3500, 'Grocery': 2500},
    'March': {'Electronics': 5500, 'Clothing': 3200, 'Grocery': 2400}
}


for months in sales:
    sales[months]['Home Appliances'] = random.randint(4000,11000)

print(sales)

total_sales = { month: sum(data.values())for month,data in sales.items()}
print('Total Sales: ',total_sales)

# Identify the best performing month by total sales
best_month = max(total_sales, key=total_sales.get)
best_sales = total_sales[best_month]

print(best_month,':',best_sales)

# Printing the top two categories that contributed the most to the sales in the best performing month.

top_categories = sorted(sales[best_month].items(),key = lambda x: x[1] ,reverse = True)[:2]
for category, value in top_categories:
    print(f"{category}: {value}")
````

### **Problem 6: Functions with Memoization and Advanced Usage**  
Write a Python program that:
1. Implements a recursive function to calculate the **Nth Fibonacci number** but uses **memoization** to optimize repeated calculations. Use a dictionary to store already computed Fibonacci numbers.
2. Implements a lambda function that checks if a number is **prime**, and uses it to filter prime numbers from a list of random integers.
3. Uses the `reduce` function to find the **greatest common divisor (GCD)** of a list of integers.

````python
def fibonacci(n, memo={}):
    # Check if the Fibonacci number is already calculated and in the memo
    if n in memo:
        return memo[n]
    
    # Base cases
    if n <= 1:
        return n

    # Recursive case with memoization
    memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2, memo)
    return memo[n]

# Example usage
n = 10
print(f"The {n}th Fibonacci number is {fibonacci(n)}")
````

````python
is_prime = lambda x: all(x % i != 0 for i in range(2, int(x**0.5) + 1)) and x > 1

is_prime(120)
````

````python
from functools import reduce
from math import gcd

def gcd_of_list(numbers):
    return reduce(gcd, numbers)

# Example usage
numbers = [48, 64, 80, 96]
print(f"The GCD of the list is: {gcd_of_list(numbers)}")
````

### **Problem 7: OOP Concepts**  
Create a class `BankAccount` that:
1. Allows a user to create an account by providing a **name** and an initial **balance**.
2. Includes methods to **deposit**, **withdraw**, and **check balance**. The withdrawal should not allow an account to go negative.
3. Overloads the `+` operator to allow merging of two accounts (sum of their balances) and returns a new account.
4. Uses **Inheritance** to create a subclass `SavingsAccount` that adds an interest rate. Implement a method to calculate interest on the balance and update it accordingly.

````python
class BankAccount:

    def __init__(self, name, balance = 0):
        self.name = name
        self.balance = balance
        
    def deposit(self, deposit=0):
        self.balance += deposit
        print(f'{deposit} amount has been credited in your account.')
        
    def withdraw(self, withdraw_amount):
        
        if self.balance < withdraw_amount:
            print('Insufficient Balance')
        else:
            self.balance -= withdraw_amount
            print(f'{withdraw_amount} amount has been debited from your account.')
        
    def checkBalance(self):
        return f' Total Balance : {self.balance} '
        
    def __add__(self,other):
        merged_account = BankAccount(f"{self.name} & {other.name}", self.balance + other.balance)
        return merged_account

    
class SavingsAccount(BankAccount):
    
    def __init__(self, name, balance=0 ):
        super().__init__(name, balance)  # Call the parent constructor
        self.interest_rate = 12  # Interest rate as a percentage (e.g., 2 for 2%)

    def calculate_interest(self):
        interest = (self.balance * self.interest_rate) / 100  # Calculate interest
        self.balance += interest  # Update balance
        print(f'Interest of {interest:.2f} has been added to your account.')
        
account1 = SavingsAccount('Namling Limbu', 200)
account1.deposit(500)
account1.withdraw(150)
account1.checkBalance()

account1.calculate_interest()
````

````python
name_and_score = {}
with open('Nameandscore.txt', 'r') as f:
    data = f.read()
    print(data)
    f.seek(0)
    for line in f:
        name,score = line.strip().split(', ')
        name_and_score[name] = score
        
sorted_name_and_score = dict(sorted(name_and_score.items(), key = lambda item : item[1], reverse = True))
print(sorted_name_and_score)
````
